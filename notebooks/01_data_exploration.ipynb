# Cognivision Voice Intelligence
## Notebook 01: Dataset Exploration & Understanding (CREMA-D)
This notebook performs comprehensive exploratory data analysis (EDA) on the CREMA-D English emotional speech dataset, inspecting audio distributions, speaker demographics, and emotion class balances.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root to sys.path
root_dir = Path("..").resolve()
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

from backend.app.services.dataset_service import dataset_service
from backend.app.utils.constants import EMOTIONS
```

### 1. Load Dataset Metadata

```python
df = dataset_service.load_metadata_if_exists()
if df is None:
    print("Scanning dataset to generate metadata...")
    dataset_service.scan_and_generate_metadata()
    df = dataset_service.load_metadata_if_exists()

print(f"Total audio files in dataset: {len(df)}")
df.head(10)
```

### 2. Dataset Summary Statistics

```python
summary = dataset_service.get_summary()
print(f"Valid Files: {summary['valid_files']}")
print(f"Corrupted Files: {summary['corrupted_files']}")
print(f"Total Speech Duration: {summary['duration_stats']['total_hours']} hours")
print(f"Mean Duration: {summary['duration_stats']['mean_seconds']} seconds")
print(f"Sample Rate: {summary['sample_rate_distribution']}")
```

### 3. Emotion Class Balance

```python
plt.figure(figsize=(9, 4.5))
emotion_counts = df['emotion'].value_counts()
sns.barplot(x=emotion_counts.index, y=emotion_counts.values, palette='viridis')
plt.title("CREMA-D Emotion Class Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Emotion")
plt.ylabel("Audio Clip Count")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### 4. Duration Distribution by Emotion

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='emotion', y='duration', palette='Set2')
plt.title("Audio Duration by Emotion Category", fontsize=14, fontweight='bold')
plt.xlabel("Emotion")
plt.ylabel("Duration (seconds)")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### 5. Speaker Demographics & Sentence Analysis

```python
print(f"Unique Actors: {df['actor_id'].nunique()}")
print(f"Unique Sentences: {df['sentence_id'].nunique()}")
actor_counts = df['actor_id'].value_counts()
print(f"Clips per Actor: Min={actor_counts.min()}, Max={actor_counts.max()}, Median={actor_counts.median()}")
```
